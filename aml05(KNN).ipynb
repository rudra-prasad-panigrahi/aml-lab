```python
import math
from collections import Counter

#1 setup
points = ['A', 'B', 'C', 'D']
X1 = [7,8,5,9]
X2 = [2,1,4,6]
labels = ['pos', 'neg','pos', 'neg']
p = [8,2]
k = 3
```

```python
#2 calculating euclidean distances
dataset = []
for i in range(len(points)):
  #formula:sqrt((x1-x2)^2 + (y1-y2)^2)
  dist = math.sqrt((p[0]-X1[i])**2 + (p[1]-X2[i])**2)
  dataset.append({'point': points[i], 'dist': dist, 'label': labels[i]})
print(dataset)
```

```python
#3 sort by distance(closest first)
dataset.sort(key=lambda x: x['dist'])
#4 get top k neighbours
neighbours = dataset[:k]

#5 voting(classify)
neighbour_labels = [n['label'] for n in neighbours]
prediction = Counter(neighbour_labels). most_common(1)[0][0]
print(prediction)
```

```python
# output Results
print(f"Target point p{p} Results:")
print("-"*30)
for n in neighbours:
  print(f"Neighbour {n['point']}: Distance {n['dist']:.2f}, Class: {n['label']}")
print("-"*30)
print(f"Prediction: {prediction}")
```

**Anova Test**

```python
import numpy as np
from scipy.stats import f
```

```python
samsung = np.array([36000, 35000, 31000, 42000, 38000])
google = np.array([44000, 45000, 51000, 50000,  52000])
sony = np.array([16000, 15000, 18000, 24000, 22000])
```

```python
mean_samsung = np.mean(samsung)
mean_google = np.mean(google)
mean_sony = np.mean(sony)

print("mean samsung=",mean_samsung)
print("mean google=",mean_google)
print("mean sony=",mean_sony)

```

```python
all_data = np.concatenate([samsung, google, sony])
grand_mean = np.mean(all_data)
print("grand mean=",grand_mean)
```

```python
n_samsung = len(samsung)
n_google = len(google)
n_sony = len(sony)

ss_between = (n_samsung * (mean_samsung - grand_mean) ** 2 +
              n_google * (mean_google - grand_mean) ** 2 +
              n_sony * (mean_sony - grand_mean) ** 2)


print("SSB=",ss_between)
```

```python
SSW_samsung = np.sum((samsung - mean_samsung) ** 2)
SSW_google = np.sum((google - mean_google) ** 2)
SSW_sony = np.sum((sony - mean_sony) ** 2)
SSW = SSW_samsung + SSW_google + SSW_sony

print("SSW samsung=",SSW_samsung)
print("SSW google=",SSW_google)
print("SSW sony=",SSW_sony)

print("SSW=",SSW)
```

```python
K =3
N = len(all_data)
df_between = K - 1
df_within = N - K

print("df_between=",df_between)
print("df_within=",df_within)
print("df Total = ",df_between + df_within)
```

```python
MSB = ss_between / df_between
MSW = SSW / df_within

print("MSB=",MSB)
print("MSW=",MSW)
```

```python
F_calculated = MSB / MSW
print("F_calculated=",F_calculated)
```

```python
alpha = 0.05
F_critical = f.ppf(1 - alpha, df_between, df_within)
print("F_critical=",F_critical)
```

```python
if F_critical >= F_calculated:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
#
```
