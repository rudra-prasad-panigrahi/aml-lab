```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn import linear_model
```

```python
df =pd.read_csv("/student_data.csv")
df
```

```python
plt.scatter(df['studied(x1)'],df['Score(y)'])
plt.xlabel("studied(x1)")
plt.ylabel("Score(y)")
plt.show()
```

```python
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(df['studied(x1)'], df['slept(x2)'], df['Score(y)'], c='blue', marker='o')

ax.set_xlabel('studied(x1)')
ax.set_ylabel('slept(x2)')
ax.set_zlabel('Score(y)')
ax.set_title('3D Scatter Plot of Score vs. Study and Sleep Time')

plt.show()
```
