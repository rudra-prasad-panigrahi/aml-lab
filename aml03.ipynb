# **logistic regression using gradient_descent**

---




```python
import numpy as np
```

```python
x1 = np.array([3,1,2,4,3])
x2 = np.array([6,3,4,7,5])
y = np.array([1,0,0,1,1])
```

```python
def gradient_descent(x1,x2,y):
  m1_curr = m2_curr = b_curr =0
  rate = 0.1
  n = len(x1)
  for i in range(10):
    y_predicted = m1_curr * x1 + m2_curr * x2 + b_curr
    cost = (1/n) * sum([val**2 for val in (y-y_predicted)])
    md1  = -(2/n)*sum(x1*(y-y_predicted))
    md2  = -(2/n)*sum(x2*(y-y_predicted))
    bd = -(2/n)*sum(y-y_predicted)
    m1_curr = m1_curr - rate * md1
    m2_curr = m2_curr - rate * md2
    b_curr = b_curr - rate *bd
    print("m1 {}, m2 {}, b {}, cost {} iteration {}".format(m1_curr,m2_curr,b_curr,cost,i))
```

```python
gradient_descent(x1,x2,y)
```
