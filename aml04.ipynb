```python
import numpy as np
import pandas as pd
from sklearn.linear_model import LinearRegression,Ridge,Lasso
from sklearn.metrics import mean_squared_error
```

```python
#x1 is more important, x2 is less important, x3 is noise
x = np.array([
    [1,2,10],
    [2,1,12],
    [3,3,9],
    [4,5,11],
    [5,4,13],
    [6,6,8],
    [7,7,7],
    [8,8,14]
])

# y mainly depends on x1 and x2
y = np.array([6,7,12,18, 20, 24, 28, 32])
```

```python
df = pd.DataFrame(x, columns=['x1','x2','x3'])
df['y'] = y
print("dataset: \n",df)
```

```python
# without regularization

lr = LinearRegression()
lr.fit(x,y)
y_pred_lr = lr.predict(x)

print("linear Regression(no Regularization)")
print("coefficients:",lr.coef_)
print("intercept:",lr.intercept_)
print("MSE:",mean_squared_error(y,y_pred_lr))
```

```python
# ridge regresion(L2)
ridge = Ridge(alpha=1.0)
ridge.fit(x,y)
y_pred_ridge = ridge.predict(x)

print("ridge Regression")
print("coefficients:",ridge.coef_)
print("intercept:",ridge.intercept_)
print("MSE:",mean_squared_error(y,y_pred_ridge))
```

```python
#Lasso regression(L1)
lasso = Lasso(alpha =0.5)
lasso.fit(x,y)
y_pred_lasso = lasso.predict(x)

print("Lasso Regression")
print("coefficients:",lasso.coef_)
print("intercept:",lasso.intercept_)
print("MSE:",mean_squared_error(y,y_pred_lasso))
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import Lasso

# Ensure x and y are defined for this context, using the x and y from cell 06UrOYsCAX_3
x = np.array([
    [1,1,50],
    [2,2,20],
    [3,3,80],
    [4,4,30],
    [5,5,60],
    [6,6,10],
    [7,7,90],
    [8,8,40]
])
y = np.array([5,10, 15, 20, 25, 30, 35, 40])

# Perform Lasso regression to get y_pred_lasso
lasso = Lasso(alpha=0.5) # Using the same alpha as in HX7rjMFdB70p
lasso.fit(x, y)
y_pred_lasso = lasso.predict(x)

# Calculate residuals for Lasso Regression
residual_lasso = y - y_pred_lasso

# Plot residuals for Lasso Regression
fig, ax = plt.subplots(figsize=(10, 6))
sns.scatterplot(x=y_pred_lasso, y=residual_lasso, ax=ax)
ax.axhline(y=0, color='r', linestyle='--')
ax.set_title('Lasso Regression: Residuals vs. Predicted Values')
ax.set_xlabel('Predicted Values')
ax.set_ylabel('Residuals')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
