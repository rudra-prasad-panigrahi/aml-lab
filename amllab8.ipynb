```python
import math
```

```python
X = [1,2,3,4,5]
Y = [40,45,50,60,65]

n = len(X)
```

```python
sumx=sum(X)
sumy=sum(Y)
meanX=sumx/n
meanY=sumy/n
print("mean of x is",meanX)
print("mean of y is",meanY)
print("sum of x is",sumx)
print("sum of y is",sumy)
```

```python
x_dev=[]
y_dev=[]
for x,y in zip(X,Y):
  dx=x-meanX
  dy=y-meanY
  x_dev.append(dx)
  y_dev.append(dy)
print("x_dev",x_dev)
print("y_dev",y_dev)
```

```python
import builtins
prd=[]
x_sq=[]
y_sq=[]
for x,y in zip(x_dev,y_dev):
  p=x*y
  x2=x**2
  y2=y**2
  prd.append(p)
  x_sq.append(x2)
  y_sq.append(y2)
sum_prd=builtins.sum(prd)
sum_x_sq=builtins.sum(x_sq)
sum_y_sq=builtins.sum(y_sq)
print("sum of product",sum_prd)
print("sum of x_sq",sum_x_sq)
print("sum of y_sq",sum_y_sq)
nume=sum_prd
dino=math.sqrt(sum_x_sq*sum_y_sq)
r=nume/dino
print("r",r)
if r>0.7:
  print("strong positive relation")
elif r>0.3 and r<0.5:
  print("weak positive relation")
elif r==0:
  print("no relation")
elif r<-0.3 and r>-0.5:
  print("weak negative relation")
elif r<-0.7:
  print("strong negative relation")
```

```python
import numpy as np
import pandas as pd
```

```python
st=["a",'b',"c","d","e","f"]
sthr=[2,4,6,8,10,12]
mrk=[50,60,80,70,90,100]
```

```python
df=pd.DataFrame({
    "st":st,
    "sthr":sthr,
    "mrk":mrk
})
print(" original data")
df
```

```python
df["rankx"]=df["sthr"].rank(
    method="average",
    ascending=True
).astype(int)
df["ranky"]=df["mrk"].rank(
    method="average",
    ascending=True
).astype(int)
print("rank data")
df
```

```python
df["d"]=df["rankx"]-df["ranky"]
df["d_sq"]=df["d"]**2
df
```

```python
sum_d2=df["d_sq"].sum()
print(sum_d2)
n=len(df)
print("no of student:",n)
```

```python
rho=1-(6*sum_d2)/(n*(n**2-1))
print("rho",rho)
```

```python
if rho>0.9:
  print("strong positive relation")
elif rho>0.5 and rho<0.9:
  print("weak positive relation")
elif rho==0:
  print("no relation")
```
