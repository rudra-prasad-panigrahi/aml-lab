```python
!pip install langchain_groq

from langchain_groq import ChatGroq
from langchain.tools import tool
from langchain.agents import create_agent
```

```python
PRODUCTS = {
    "wireless headphones": {"price": 79.99, "rating": 4.6, "discription": "Over ear bluetooth, 30hrs battery, ANC"},
    "smart watch": { "price": 199.99, "rating": 4.3, "discription": "Tracks heart rate and sleep, 5-day battery, water resistant"},
    "mechanical keyboard": {"price": 129.99, "rating": 4.8, "discription": "Tenkeyles, Cherry MX Brown switches, per key RGB"},
    "laptopstand": {"price": 23.99, "rating": 4.5, "discription": "Adjustable aluminium, fits 11-17 inch lapto[, folds"}
}
@tool
def get_product(name: str) -> str:
    """Look up a product by name and return its price, rating and discription"""
    p = PRODUCTS.get(name.lower())
    if not p:
        return f"Product not found. Available {', '.join(PRODUCTS.keys())}"
    return str(p)
```

```python
from google.colab import userdata
x = userdata.get('GORK_API_KEY')
```

```python
llm = ChatGroq(model = "openai/gpt-oss-120b", api_key = x)
pmt = "You are a helpful product assistant for an online tech store."
agent = create_agent(
    llm,
    tools=[get_product],
    system_prompt=pmt,
)
```

```python
def ask(question: str):
    result = agent.invoke({"messages": [{"role": "user",
                                       "content": question}]})
    print(result["messages"][-1].content)
```

```python
ask("what is the price of wireless headphones.")
```

```python
REVIEWS = {
    "wireless headphones": {"reviews": 1262, "rating": 4.6},
    "smart watch": {"reviews": 1262, "rating": 4.3},
    "mechanical keyboard": {"reviews": 1262, "rating": 4.8},
    "laptopstand": {"reviews": 1262, "rating": 4.5}
}
@tool
def get_review(name: str) -> str:
    """Look up a product review by a product name.
       Return the product name, number of reviews and rating"""
    r = REVIEWS.get(name.lower())
    if not r:
        return f"Reviews not available for this product"
    return str(r)
```

```python
agent2 = create_agent(
    llm,
    tools = [get_product, get_review],
    system_prompt= "You are a helpful product assistant for an online tech store.",
)
def ask2(question: str):
    result = agent2.invoke({"messages": [{"role": "user",
                                       "content": question}]})
    print(result["messages"][-1].content)
```

```python
ask2("how do people like smart watch")
```

```python
from langgraph.checkpoint.memory import InMemorySaver
checkpointer = InMemorySaver()
pmt = "You are a helpful product assistant for an online tech store."
agent = create_agent(
    llm,
    tools=[get_product],
    system_prompt=pmt,
    checkpointer = checkpointer
)
def ask2(question: str, thread_id: str = "usser_anmol_session-1"):
    config = {"configurable": {"thread_id": thread_id}}
    result = agent.invoke({"messages": [{"role": "user",
                                       "content": question}],
                           config: config})
    print(result["messages"][-1].content)
```

```python
ask2("how do people like smart watch")
```
