```python
import numpy as np
import pandas as pd
from scipy.stats import f
```

```python
data = np.array([
    [70,72,68,75],
    [65,67,70,69],
    [80,78,82,81]
], dtype = float)

methods = ["Method A", "Method B", "Method C"]
semesters = ["sem 1", "sem 2", "sem 3", "sem 4"]

df = pd.DataFrame(data, index = methods, columns = semesters)
print("Observe Data")
display(df)

print("\n Row totals")
print(df.sum(axis = 1))

print("\n Column totals")
print(df.sum(axis = 0))
```

```python
r,c = data.shape
N= r*c

grand_total =data.sum()
grand_mean=grand_total /N

print(f"Number of rows(Teaching method) : {r}")
print(f"Number of columns(semester): {c} ")
print(f"Number of Observations: {N}")
print(f"Grand Total: {grand_total}")
print(f"Grand Mean: {grand_mean}")
```

```python
sum_of_squares = np.sum(data**2)
correction_factor = grand_total ** 2/N
sum_of_squares_corrected = sum_of_squares - correction_factor
print(f"Sum of Squares: {sum_of_squares}")
print(f"Sum of Squares Corrected: {sum_of_squares_corrected}")
print(f"Correction Factor: {correction_factor}")
```

```python
row_totals = data.sum(axis= 1)
ss_methods = np.sum(row_totals ** 2) / c- correction_factor
print(f"Row Totals: {row_totals}")

for method, total in zip(methods, row_totals):
  print(f"{method}: {total}")

print(f"\n ss_methods = {ss_methods}")
```

```python
column_totals = data.sum(axis =0)
ss_semesters = np.sum(column_totals ** 2) / r - correction_factor
print(f"Column Totals: {column_totals}")

for semesters, total in zip(semesters, column_totals):
  print(f"{semesters}: {total}")

print(f"\n ss_semesters = {ss_semesters}")
```

```python
ss_error = sum_of_squares_corrected - ss_methods - ss_semesters
print(f"Sum of Squares Corrected: {sum_of_squares_corrected}")
print(f"ss_error = {ss_error}")
```
